# Tools

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#x_train is the input variable (size in 1000 sq ft)
#y_train is the output variable (price in 1000s of dollars)

x_train = np.array([1.0, 2.0])
y_train = np.array([300.0 , 500.0])
print("x_train:", x_train)
print("y_train:", y_train)
```

```python
# m is the number of training examples
m = x_train.shape[0]
print("Number of training examples (m):", m)
```

```python
i = 0
x_i = x_train[i]
y_i = y_train[i]
print(f"Example {i}: x = {x_i}, y = {y_i}")
```

```python
#Plotting the data points
plt.scatter(x_train, y_train, marker = 'x', c = 'r')
plt.title('Housing Prices')
plt.ylabel('Price (in 1000s of dollars)')
plt.xlabel('Size (in 1000 sq ft)')
plt.show()
```

```python
w = 100
b = 100
print(f"w: {w}, b: {b}")
```

```python
def compute_model(x, w, b):
    m = x.shape[0]
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w * x[i] + b
    return f_wb
```

```python
tmp_f_wb = compute_model(x_train, w, b)
print("Computed model values:", tmp_f_wb)

plt.plot(x_train, tmp_f_wb, c = 'b', label = 'Our prediction')
plt.scatter(x_train, y_train, marker = 'x', c = 'r', label = 'Acutal Values')
plt.title('Housing Prices')
plt.ylabel('Price (in 1000s of dollars)')
plt.xlabel('Size (in 1000 sq ft)')
plt.legend()
plt.show()
```

```python
w = 200
b = 100
print(f"w: {w}, b: {b}")

tmp_f_wb = compute_model(x_train, w, b)
print("Computed model values:", tmp_f_wb)

plt.plot(x_train, tmp_f_wb, c = 'b', label = 'Our prediction')
plt.scatter(x_train, y_train, marker = 'x', c = 'r', label = 'Acutal Values')
plt.title('Housing Prices')
plt.ylabel('Price (in 1000s of dollars)')
plt.xlabel('Size (in 1000 sq ft)')
plt.legend()
plt.show()
```
